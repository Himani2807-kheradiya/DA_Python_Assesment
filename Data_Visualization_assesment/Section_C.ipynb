```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
np.random.seed(7)
n = 200

df = pd.DataFrame({
    "restaurant": np.random.choice(
        ["Pizza Hub", "Burger Point", "Spice House",
         "Food Corner", "Tasty Bites"], n
    ),
    "order_value": np.random.uniform(100, 800, n),
    "distance_km": np.random.uniform(1, 20, n),
    "delivery_time": np.random.normal(30, 8, n),
    "rating": np.random.uniform(2.5, 5.0, n)
})

df.loc[0, "order_value"] = np.nan
df.loc[1, "delivery_time"] = np.nan

numeric_cols = df.select_dtypes(include="number").columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

print("Dataset shape:", df.shape)
print("Total missing values:", df.isnull().sum().sum())
df.head()
```

```python
def summary_statistics():
    print("\\n--- Summary Statistics ---")
    for col in df.select_dtypes(include="number").columns:
        print(
            f"{col}: mean = {np.mean(df[col]):.2f}, "
            f"standard deviation = {np.std(df[col]):.2f}"
        )

    df.describe().to_csv("summary_statistics.csv")
    print("Saved: summary_statistics.csv")


def distribution_analysis():
    fig, ax = plt.subplots(1, 2, figsize=(10, 4))

    ax[0].hist(df["order_value"], bins=10)
    ax[0].set_title("Order Value Distribution")
    ax[0].set_xlabel("Order Value (Rs)")
    ax[0].set_ylabel("Number of Orders")

    ax[1].hist(df["delivery_time"], bins=10)
    ax[1].set_title("Delivery Time Distribution")
    ax[1].set_xlabel("Delivery Time (minutes)")
    ax[1].set_ylabel("Number of Orders")

    plt.tight_layout()
    plt.savefig("distribution_analysis.png", dpi=150)
    plt.close(fig)
    print("Saved: distribution_analysis.png")



def correlation_heatmap():
    corr = df.select_dtypes(include="number").corr()

    fig, ax = plt.subplots(figsize=(7, 5))
    sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f", ax=ax)
    ax.set_title("Food Delivery Correlation Heatmap")
    plt.tight_layout()
    plt.savefig("correlation_heatmap.png", dpi=150)
    plt.close(fig)
    print("Saved: correlation_heatmap.png")



def restaurant_performance():
    report = df.groupby("restaurant").agg(
        mean_rating=("rating", "mean"),
        total_orders=("rating", "count"),
        avg_order_value=("order_value", "mean")
    ).round(2)

    print("\\n--- Restaurant Performance ---")
    print(report)

    fig, ax = plt.subplots(figsize=(8, 5))
    report["mean_rating"].sort_values(ascending=False).plot(
        kind="bar", ax=ax, title="Average Restaurant Ratings"
    )
    ax.set_xlabel("Restaurant")
    ax.set_ylabel("Mean Rating")
    plt.tight_layout()
    plt.savefig("restaurant_performance.png", dpi=150)
    plt.close(fig)
    print("Saved: restaurant_performance.png")


def final_report():
    print("\\n========== FINAL EDA REPORT ==========")

    top3 = df.groupby("restaurant")["rating"].mean()
    print("\\nTop 3 Restaurants by Mean Rating:")
    print(top3.sort_values(ascending=False).head(3).round(2).to_string())

    corr = df.select_dtypes(include="number").corr()
    upper = corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
    correlations = upper.abs().stack()

    if not correlations.empty:
        pair = correlations.idxmax()
        print("\\nNumeric pair with highest absolute Pearson correlation:")
        print(f"{pair[0]} and {pair[1]}: r = {corr.loc[pair[0], pair[1]]:.3f}")
    else:
        print("\\nNot enough numeric columns to calculate correlation.")

    print(f"\\nMean delivery time: {df['delivery_time'].mean():.2f} minutes")
    print(f"Standard deviation of delivery time: {df['delivery_time'].std():.2f} minutes")
```

```python
while True:
    print("===== FOOD DELIVERY ANALYTICS =====")
    print("1. Summary Statistics")
    print("2. Distribution Analysis")
    print("3. Correlation Heatmap")
    print("4. Restaurant Performance Report")
    print("5. Exit")

    choice = input("Enter your choice (1-5): ").strip()

    if choice == "1":
        summary_statistics()
    elif choice == "2":
        distribution_analysis()
    elif choice == "3":
        correlation_heatmap()
    elif choice == "4":
        restaurant_performance()
    elif choice == "5":
        final_report()
        print("\\nThank you for using Food Delivery Analytics!")
        break
    else:
        print("Invalid choice. Please enter a number from 1 to 5.")
```
